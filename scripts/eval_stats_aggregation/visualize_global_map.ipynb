```python
import h5py
import matplotlib.pyplot as plt
import os
```

```python
runs_path = "/vision/asomaya1/active-rir/runs/"
model = " global_map"


h5py_save_path = os.path.join(runs_path, model, "tb", "visualized_maps.h5")

# Open the H5 file
h5file = h5py.File(h5py_save_path, "r")

# Access the dataset containing the maps
maps_dataset = h5file["maps"]

# Iterate through the maps and visualize them
for episode_map in maps_dataset:
    plt.imshow(episode_map, cmap='viridis')  # You can choose a different colormap
    plt.colorbar()
    plt.show()

# Close the H5 file
h5file.close()
```

```python
import numpy as np
runs_path = "/vision/asomaya1/active-rir/runs/"
model = "test_coverage_reward"


path = os.path.join(runs_path, model, "tb", "global_map.npy")
```

```python
data = np.load(path)
```

```python
data.shape
```

```python
# Extracting the two channels
channel_1 = data[0, 0, :, :]
channel_2 = data[0, 1, :, :]

# Plotting the two channels
fig, axes = plt.subplots(1, 2, figsize=(12, 6))

axes[0].imshow(channel_1, cmap='gray')
axes[0].set_title('Channel 1')

axes[1].imshow(channel_2, cmap='gray')
axes[1].set_title('Channel 2')

plt.tight_layout()
plt.show()
```
